```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain.chat_models import init_chat_model

load_dotenv()

chat = ChatGroq(model="llama-3.1-8b-instant", api_key=os.getenv("GROQ_API_KEY"),streaming=True)

# streaming response (chunks of data as they become available)

for chunk in chat.stream("What is the capital of India?"):
    print(chunk.content, end="", flush=True)
```
